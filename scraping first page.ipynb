```python
import requests
from bs4 import BeautifulSoup
import csv
import time
```

```python
base_urls = [
    "https://wuzzuf.net/search/jobs/?a=navbg%7Cspbg&q=business%20data%20analyst&start=",
    "https://wuzzuf.net/search/jobs/?q=data+analyst&start="
]
```

```python
with open("wuzzuf_jobs_first_page.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Job Title", "Company Name", "Location", "Job Type", "Job Link"])

    for base_url in base_urls:
        print(f"Scraping jobs from: {base_url}")
        url = base_url
        response = requests.get(url)
        soup = BeautifulSoup(response.content, "html.parser")
        job_blocks = soup.find_all("div", class_="css-pkv5jc")
        for job in job_blocks:
             try:
                 job_title = job.find('h2', class_='css-m604qf').text.strip()
                 company_name = job.find('a', class_='css-17s97q8').text.strip()
                 company_location = job.find('span', class_='css-5wys0k').text.strip()
                 job_type = job.find('div', class_='css-1lh32fc').text.strip()
                 link = "https://wuzzuf.net" + job.find('a', class_="css-o171kl")['href'].strip()

                 writer.writerow([job_title, company_name, company_location, job_type, link])
             except Exception as e:
                    print(f"    Skipped a job block due to error: {e}")

        time.sleep(1)  # Pause between pages to avoid getting blocked
```
